## 課題2

「スイカ割りゲーム」のプログラムをクラスの形に整理してください。

![](https://qiita-user-contents.imgix.net/https%3A%2F%2Fqiita-image-store.s3.ap-northeast-1.amazonaws.com%2F0%2F510221%2Fcf759119-d591-ebdd-22e1-b42e840e91c0.png?ixlib=rb-4.0.0&auto=format&gif-q=60&q=75&w=1400&fit=max&s=a8d6af0665717fe07630cabc3c2f301f)

````python
import random
import math

BOARD_SIZE = 5  # ボードサイズを定義

def generate_position(board_size):
    # スイカとプレイヤーの初期位置をランダムに生成
    return (random.randrange(0, board_size), random.randrange(0, board_size))

def determine_move(player_pos, target_pos):
    # スイカに向かってプレイヤーを移動させる方向を決定
    current_x, current_y = player_pos
    target_x, target_y = target_pos

    move_x = 1 if current_x < target_x else -1 if current_x > target_x else 0
    move_y = 1 if current_y < target_y else -1 if current_y > target_y else 0

    return (move_x, move_y)

def move_player(player_pos, move_x, move_y):
    # プレイヤーの位置を更新
    new_x = player_pos[0] + move_x
    new_y = player_pos[1] + move_y
    return (new_x, new_y)

def play_game(board_size):
    suika_pos = generate_position(board_size)
    player_pos = generate_position(board_size)

    while suika_pos != player_pos:
        move_x, move_y = determine_move(player_pos, suika_pos)
        player_pos = move_player(player_pos, move_x, move_y)
        print("プレイヤーが移動しました:", player_pos)

    print("スイカを見つけました！")

# ゲームを開始
play_game(BOARD_SIZE)
````

- 属性
    - ```board_size```
    - ```suika_pos```
    - ```player_pos```
- メソッド
    - ```calc_distance```: 二点間のユークリッド距離を計算
    - ```move_player```: プレイヤーをスイカに向かって一歩移動させます。プレイヤーの位置を更新します。
    - ```play_game```: ゲームのループを実行します。プレイヤーがスイカの位置に到達するまで、プレイヤーを移動させます。

````python
import random
import math

class SuikaGame:
    def __init__(self, board_size):
        """
        SuikaGameクラスのコンストラクタ。
        ボードサイズ、スイカとプレイヤーの初期位置を設定します。
        """
        self.board_size = board_size
        self.suika_pos = self._generate_position()
        self.player_pos = self._generate_position()

    def _generate_position(self):
        """
        ボード内でランダムな位置を生成する内部メソッド。
        """
        return (random.randrange(0, self.board_size), random.randrange(0, self.board_size))

    def calc_distance(self, pos1, pos2):
        """
        二点間のユークリッド距離を計算します。
        """
        diff_x = pos1[0] - pos2[0]
        diff_y = pos1[1] - pos2[1]
        return math.sqrt(diff_x**2 + diff_y**2)

    def move_player(self):
        """
        プレイヤーをスイカに向かって一歩移動させ、プレイヤーの位置を更新します。
        """
        current_x, current_y = self.player_pos
        target_x, target_y = self.suika_pos

        move_x = 0
        if current_x < target_x:
            move_x = 1
        elif current_x > target_x:
            move_x = -1

        move_y = 0
        if current_y < target_y:
            move_y = 1
        elif current_y > target_y:
            move_y = -1

        new_x = current_x + move_x
        new_y = current_y + move_y
        self.player_pos = (new_x, new_y)

    def play_game(self):
        """
        ゲームのループを実行します。
        プレイヤーがスイカの位置に到達するまで、プレイヤーを移動させます。
        """
        print(f"スイカの初期位置: {self.suika_pos}")
        print(f"プレイヤーの初期位置: {self.player_pos}")

        while self.suika_pos != self.player_pos:
            self.move_player()
            distance = self.calc_distance(self.player_pos, self.suika_pos)
            print(f"プレイヤーが移動しました: {self.player_pos} (スイカまでの距離: {distance:.2f})")

        print("スイカを見つけました！")
````

````python
game = SuikaGame(board_size=10) # ボードサイズを5としてゲームを初期化
game.play_game()
````

````python
import random
import math

class WatermelonHunt:
    def __init__(self, board_size=5, player_num=2):
        self.board_size = board_size
        # player_numを最初に初期化
        self.player_num = player_num 
        self.suika_pos = (random.randrange(0, self.board_size), random.randrange(0, self.board_size))
        # player_numを使ってプレイヤーの初期位置を生成
        self.player_positions = [(random.randrange(0, self.board_size), random.randrange(0, self.board_size)) for _ in range(self.player_num)]
        self.winner = None
        self.turn_count = 0 # ターン数を追跡するための属性を追加

    def calc_distance(self, pos1, pos2):
        """
        二点間のユークリッド距離を計算します。
        """
        diff_x = pos1[0] - pos2[0]
        diff_y = pos1[1] - pos2[1]
        return math.sqrt(diff_x**2 + diff_y**2)

    def move_player(self, player_index):
        """
        指定されたプレイヤーをスイカに向かって一歩移動させます。
        プレイヤーの位置を更新し、スイカに到達したか確認します。
        """
        # すでにスイカに到達しているプレイヤーは移動させない
        if self.player_positions[player_index] == self.suika_pos:
            return

        current_x, current_y = self.player_positions[player_index]
        target_x, target_y = self.suika_pos

        if current_x < target_x:
            current_x += 1
        elif current_x > target_x:
            current_x -= 1

        if current_y < target_y:
            current_y += 1
        elif current_y > target_y:
            current_y -= 1

        self.player_positions[player_index] = (current_x, current_y)
        distance = self.calc_distance(self.player_positions[player_index], self.suika_pos)
        print(f"プレイヤー {player_index + 1} が移動しました: {self.player_positions[player_index]} (スイカまでの距離: {distance:.2f})")
        
        # スイカに到達したかチェック
        if self.player_positions[player_index] == self.suika_pos:
            self.winner = player_index + 1

    def play_game(self):
        """
        ゲームのメインループを実行します。
        プレイヤーがスイカの位置に到達するまで、全プレイヤーが各ターンで移動します。
        """
        print(f"スイカの初期位置: {self.suika_pos}")
        for i, pos in enumerate(self.player_positions):
            print(f"プレイヤー {i + 1} の初期位置: {pos}")

        while self.winner is None:
            self.turn_count += 1
            print(f"\n--- ターン {self.turn_count} ---")
            for i in range(self.player_num):
                # 誰かが勝っていたら、そのターンの残りのプレイヤーは移動させない
                if self.winner is not None:
                    break 
                self.move_player(i)
        
        print(f"\n--- ゲーム終了 ---")
        print(f"🎉 プレイヤー {self.winner} がスイカを見つけました！🎉")
````

````python
game = WatermelonHunt(board_size=5, player_num=3) # ボードサイズ5、プレイヤー数3でゲームを初期化
game.play_game()
````
