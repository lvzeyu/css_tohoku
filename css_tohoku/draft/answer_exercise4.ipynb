```python
import random

class SuikaWariGame:
    def __init__(self, board_size):
        self.board_size = board_size
        self.suika_pos = self.generate_position() # Changed to public
        self.player_pos = self.generate_position() # Changed to public

    def generate_position(self):
        # Generate random initial positions for the watermelon and player
        return (random.randrange(0, self.board_size), random.randrange(0, self.board_size))

    def determine_move(self):
        # Determine the direction to move the player towards the watermelon
        current_x, current_y = self.player_pos
        target_x, target_y = self.suika_pos

        move_x = 0
        if current_x < target_x:
            move_x = 1
        elif current_x > target_x:
            move_x = -1

        move_y = 0
        if current_y < target_y:
            move_y = 1
        elif current_y > target_y:
            move_y = -1

        return (move_x, move_y)

    def move_player(self, move_x, move_y): # Changed to public
        # Update the player's position
        new_x = self.player_pos[0] + move_x
        new_y = self.player_pos[1] + move_y
        self.player_pos = (new_x, new_y)

    def play_game(self):
        print("スイカ割りゲームを開始します！")
        print(f"スイカの初期位置: {self.suika_pos}")
        print(f"プレイヤーの初期位置: {self.player_pos}\n")

        while self.suika_pos != self.player_pos:
            move_x, move_y = self.determine_move()
            self.move_player(move_x, move_y) # Changed to public
            print(f"プレイヤーが移動しました: {self.player_pos}")

        print("\nスイカを見つけました！ゲームクリア！")

# ゲームを開始
BOARD_SIZE = 5
game = SuikaWariGame(BOARD_SIZE)
game.play_game()
```
