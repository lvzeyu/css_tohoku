## print文


### print文の基本


```print()```関数の引数には数値、配列、タプル、オブジェクトなどを指定することができます。


`````python
print(3)                   #=> 3
print([1, 2, 3])           #=> [1, 2, 3]
print((1, 2, 3))           #=> (1, 2, 3)
print({'k1':10, 'k2':20})  #=> {'k2': 20, 'k1': 10}
`````

引数をカンマ(```,```)で複数指定すると、ひとつのスペースで区切って複数の値を表示します。

`````python
print("AAA", "BBB", 30)         #=> AAA BBB
`````

### フォーマットされた出力

より高度的な使い方として、出力結果をよりわかりやすく整形することができます。

#### ```.format()```

```.format()```メソッドを使う基本的な構文は、中括弧```{}```を使って文字列内にプレースホルダーを設定し、それに```.format()```で値を渡す形です。

`````python
name = "山田"
age = 25
print("名前：{} 年齢：{}".format(name, age))  #=> 名前：山田 年齢：25
`````

プレースホルダーに位置を指定することもできます。これにより、引数の順序を自由に変更したり、同じ値を複数回使用したりできます。

`````python
name = "山田"
age = 25
print("名前：{1} 年齢：{0}".format(age, name))  #=> 名前：山田 年齢：25
`````

キーワード引数を使って値を指定することができます。

`````python
print("名前：{name}、職業：{job}".format(name="木村", job="学生"))
`````

#### f-strings

Python 3.6以降では、文字列内で変数を直接埋め込むためのf-stringsが導入されました。

f-stringを使用するには、文字列の前に```f```または```F```を置き、波括弧```{}```の中に変数名や式を書きます。

`````python
name = "山田"
age = 25
print(f"名前：{name}、年齢：{age}")
`````

f-stringsでは、値の書式を指定することもできます。これは、値を特定の形式（例えば、小数点以下の桁数を指定する）で表示したい場合に便利です。

`````python
# 小数点以下の桁数を指定
pi = 3.14159265359
print(f"円周率は約 {pi:.2f}")  # 円周率は約 3.14
`````

`````python
# パーセンテージ表示
y = 0.12345
print(f"{y:.4%}") # 12.3450% (パーセンテージ表示)
`````

## uvを用いて仮想環境の構築

uvは、Rustで実装された高速なPythonパッケージ・プロジェクト管理ツールです。従来のツールよりも処理が速く、最大でpipの10〜100倍の速度でインストールできます。

### uvのインストール

```
# MacやLinuxの場合
curl -LsSf https://astral.sh/uv/install.sh | sh

# macOS は Homebrew でも可

brew install uv

# Windowsの場合はコマンドプロンプトまたはPowerShellで下記を実行
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"

```
Windowsの場合では、環境変数を設定する必要があるかもしれません、詳細は[こちらの記事](https://note.com/kohaku935/n/ne405a010682a#588082d4-96c5-4701-9452-9359807ff925)を参照しなさい。


uvがインストールされたかどうかを確認します。

```
uv --version
```

### プロジェクトの作成

```
# プロジェクトフォルダを作成して移動
mkdir my-uv-project
cd my-uv-project
```

```
# プロジェクトを初期化
uv init
```

プロジェクトを作成すると、下記のファイルが自動で作成されます。

![](./Figure/uv.png)

### 仮想環境の作成

プロジェクトごとに仮想環境を作成することが可能です。

```
# 仮想環境を作成
uv venv
```

コマンドを実行すると、`.venv`という名前の仮想環境が作成されます。

`source .venv/bin/activate`スクリプトを実行すると、プロンプトの先頭にプロジェクト名が表示され、仮想環境が有効になったことが確認できます。
