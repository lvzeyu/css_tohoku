## 課題1

「スイカ割りゲーム」のプログラムを作成する。

- スイカとプレイヤーの初期位置をランダムに設定
- プレイヤーがスイカの位置に到達するまでに移動を続けて、移動する度にプレイヤーとスイカの距離を計算する
- スイカに向かってプレイヤーを移動させる方向を決定する

````python
import random
import math

BOARD_SIZE = 5  # ボードサイズを定義

def calc_distance(pos1, pos2):
    # 二点間のユークリッド距離を計算
    diff_x = pos1[0] - pos2[0]
    diff_y = pos1[1] - pos2[1]
    return math.sqrt(diff_x**2 + diff_y**2)

# スイカとプレイヤーの初期位置をランダムに設定
suika_pos = (random.randrange(0, BOARD_SIZE), random.randrange(0, BOARD_SIZE))
player_pos = (random.randrange(0, BOARD_SIZE), random.randrange(0, BOARD_SIZE))

# プレイヤーがスイカの位置に到達するまで処理を続ける
while (suika_pos != player_pos):
    # スイカまでの距離を表示
    distance = calc_distance(suika_pos, player_pos)
    print("スイカまでの距離:", distance)

    # スイカに向かってプレイヤーを移動させる方向を決定
    current_x, current_y = player_pos
    target_x, target_y = suika_pos

    if current_x < target_x:
        current_x += 1
    elif current_x > target_x:
        current_x -= 1

    if current_y < target_y:
        current_y += 1
    elif current_y > target_y:
        current_y -= 1

    # プレイヤーの位置を更新
    player_pos = (current_x, current_y)
    print("プレイヤーが移動しました:", player_pos)

print("スイカを見つけました！")
````

このプログラムを関数で分割して、再作成してください。

- 初期位置の作成の関数```generate_position()```
- 移動先を決める関数
- 移動処理の関数```move_player()```
- ゲームの主なループを実行する関数```play_game()```

````python
import random
import math

BOARD_SIZE = 5  # Define the board size

def generate_position(board_size):
    """
    Generates a random (x, y) position within the board boundaries.
    """
    return (random.randrange(0, board_size), random.randrange(0, board_size))

def determine_move_direction(player_pos, target_pos):
    """
    Determines the direction (dx, dy) the player should move to get closer to the target.
    Returns (0, 0) if already at the target.
    """
    current_x, current_y = player_pos
    target_x, target_y = target_pos

    move_x = 0
    if current_x < target_x:
        move_x = 1
    elif current_x > target_x:
        move_x = -1

    move_y = 0
    if current_y < target_y:
        move_y = 1
    elif current_y > target_y:
        move_y = -1

    return (move_x, move_y)

def move_player(player_pos, move_x, move_y):
    """
    Updates the player's position based on the determined move direction.
    """
    new_x = player_pos[0] + move_x
    new_y = player_pos[1] + move_y
    return (new_x, new_y)

def calc_distance(pos1, pos2):
    """
    Calculates the Euclidean distance between two points.
    """
    diff_x = pos1[0] - pos2[0]
    diff_y = pos1[1] - pos2[1]
    return math.sqrt(diff_x**2 + diff_y**2)

def play_game(board_size):
    """
    Executes the main game loop for the Suika Game.
    The player moves towards the watermelon until it's found.
    """
    suika_pos = generate_position(board_size)
    player_pos = generate_position(board_size)

    print(f"スイカの初期位置: {suika_pos}")
    print(f"プレイヤーの初期位置: {player_pos}")

    # Continue moving until the player reaches the watermelon
    while suika_pos != player_pos:
        # Display the distance to the watermelon
        distance = calc_distance(suika_pos, player_pos)
        print(f"スイカまでの距離: {distance:.2f}") # Format to 2 decimal places

        # Determine the move direction and update player's position
        move_x, move_y = determine_move_direction(player_pos, suika_pos)
        player_pos = move_player(player_pos, move_x, move_y)
        print(f"プレイヤーが移動しました: {player_pos}")

    print("スイカを見つけました！")
````

## 課題2

複数のプレイヤーが参加し、最初にスイカに到達したプレイヤーが勝つゲームを想定してください。

各プレイヤーはそれぞれ独立してスイカに向かって移動し、ゲームの進行は各ターンで全プレイヤーが一度ずつ移動する。

あるプレイヤーがスイカに到達したら、ゲームは終了します。

このようなシナリオを実装してください。


````python
import random
import math

BOARD_SIZE = 5  # ボードサイズを定義

def generate_position(board_size):
    """
    ボード内でランダムな位置を生成します。
    """
    return (random.randrange(0, board_size), random.randrange(0, board_size))

def determine_move_direction(player_pos, target_pos):
    """
    プレイヤーがターゲットに近づくために移動すべき方向 (dx, dy) を決定します。
    ターゲットにすでに到達している場合は (0, 0) を返します。
    """
    current_x, current_y = player_pos
    target_x, target_y = target_pos

    move_x = 0
    if current_x < target_x:
        move_x = 1
    elif current_x > target_x:
        move_x = -1

    move_y = 0
    if current_y < target_y:
        move_y = 1
    elif current_y > target_y:
        move_y = -1

    return (move_x, move_y)

def move_player(player_pos, move_x, move_y):
    """
    決定された移動方向に基づいてプレイヤーの位置を更新します。
    """
    new_x = player_pos[0] + move_x
    new_y = player_pos[1] + move_y
    return (new_x, new_y)

def calc_distance(pos1, pos2):
    """
    二点間のユークリッド距離を計算します。
    """
    diff_x = pos1[0] - pos2[0]
    diff_y = pos1[1] - pos2[1]
    return math.sqrt(diff_x**2 + diff_y**2)

def play_multiplayer_game(board_size, num_players):
    """
    複数のプレイヤーが参加するスイカ割りゲームのメインループを実行します。
    最初にスイカに到達したプレイヤーが勝者となります。
    """
    suika_pos = generate_position(board_size)
    
    # 各プレイヤーの初期位置を生成し、リストに格納
    players = []
    for i in range(num_players):
        players.append({
            'id': i + 1,
            'pos': generate_position(board_size)
        })

    print(f"スイカの初期位置: {suika_pos}")
    for player in players:
        print(f"プレイヤー{player['id']} の初期位置: {player['pos']}")

    game_over = False
    turn = 0
    while not game_over:
        turn += 1
        print(f"\n--- ターン {turn} ---")

        for player in players:
            if player['pos'] == suika_pos:
                continue # すでにスイカに到達しているプレイヤーはスキップ

            distance = calc_distance(player['pos'], suika_pos)
            print(f"プレイヤー{player['id']} の現在位置: {player['pos']} (スイカまでの距離: {distance:.2f})")

            move_x, move_y = determine_move_direction(player['pos'], suika_pos)
            player['pos'] = move_player(player['pos'], move_x, move_y)
            print(f"プレイヤー{player['id']} が移動しました: {player['pos']}")

            if player['pos'] == suika_pos:
                print(f"🎉 プレイヤー{player['id']} がスイカを見つけました！🎉")
                game_over = True
                break # スイカを見つけたプレイヤーがいる場合、このターンの処理を終了

        if not game_over:
            # 全てのプレイヤーが移動した後もゲームが終了していない場合
            pass # 次のターンへ

    print("\n--- ゲーム終了 ---")
````

````python
num_players = 3  # 参加するプレイヤーの数を設定
play_multiplayer_game(BOARD_SIZE, num_players)
````
